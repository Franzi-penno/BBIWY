## Lokale LLMs mit Python nutzen

```python
MODEL = "llama3.1:8b" # Name des Modells das Ollma kennt
```

```python
from ollama import chat, generate
```

```python
message = [{ 
    "role": "user",
    "content": "Hallo, wer bist du?"
}]

chat(MODEL,  message)
```

```python
response = generate(MODEL, message)
print(response)
```
