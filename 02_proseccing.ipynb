```python
def clean_stopwords(text: str, language: str = "german"):
    words = text.split(" ")
    clean_text = []
    for words in words: 
        if words not in stopwords.words(language):
            clean_text.append(words)
    return " ".join(clean_text)

# stopwords.words("german")
```

```python
text = "Hallo Welt wie geht es dir"
clean_stopwords(text, language="german")
```
